```python
from pyspark.sql.functions import lower, col

def search_jobs(query):
    return (
        jobs
        .filter(
            lower(col("search_text"))
            .contains(query.lower())
        )
        .select(
            "job_id",
            "title",
            "company",
            "location",
            "salary_max"
        )
    )
```

```python
from pyspark.sql import Row
from pyspark.sql.functions import current_timestamp

def save_job(user_id, job_id):
    df = spark.createDataFrame([
        Row(
            user_id=user_id,
            job_id=job_id
        )
    ]).withColumn(
        "saved_at",
        current_timestamp()
    )

    (
        df.write
        .mode("append")
        .saveAsTable(
            "career_os.user.saved_jobs"
        )
    )

    return f"Saved {job_id}"
```

```python
save_job(
    "demo_user",
    "5827000231"
)
```

```python
from pyspark.sql import Row
from pyspark.sql.functions import current_timestamp

def apply_job(user_id, job_id):
    df = spark.createDataFrame([
        Row(
            user_id=user_id,
            job_id=job_id,
            status="Applied"
        )
    ]).withColumn(
        "applied_at",
        current_timestamp()
    )

    (
        df.write
        .mode("append")
        .saveAsTable(
            "career_os.user.applications"
        )
    )

    return "Application recorded"
```
